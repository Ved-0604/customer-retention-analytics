# 03. RFM Customer Segmentation
Computes Recency, Frequency, and Monetary Value, applies quintile scoring (1-5), and assigns 8 distinct behavioral tiers.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.analysis import calculate_rfm_segments

features = pd.read_csv('../data/processed/customer_features.csv')
rfm_df = calculate_rfm_segments(features)

seg_summary = rfm_df.groupby('customer_segment').agg(
    customers=('customer_id', 'count'),
    total_revenue=('lifetime_net_revenue', 'sum'),
    avg_spend=('lifetime_net_revenue', 'mean'),
    avg_orders=('lifetime_orders', 'mean'),
    avg_recency=('recency_days', 'mean')
).reset_index().sort_values('total_revenue', ascending=False)
seg_summary
```

```python
plt.figure(figsize=(12, 5))
sns.barplot(data=seg_summary, x='customer_segment', y='total_revenue', palette='Blues_r')
plt.title('Total Revenue by RFM Customer Segment')
plt.xticks(rotation=30)
plt.ylabel('Total Net Revenue ($)')
plt.show()
```
