# 04. Customer Churn & Inactivity Driver Analysis
Defines 90-day inactivity churn, evaluates churn hazard across discount dependency, cancellations, channels, and fits an interpretable driver model.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.analysis import analyze_churn_drivers

features = pd.read_csv('../data/processed/customer_features.csv')
drivers = analyze_churn_drivers(features)
print(drivers)
```

```python
plt.figure(figsize=(10, 4))
sns.barplot(data=drivers, x='odds_ratio', y='feature', palette='mako')
plt.axvline(1.0, color='red', linestyle='--', label='Neutral Hazard (OR=1.0)')
plt.title('Churn Driver Odds Ratios (Logistic Regression)')
plt.xlabel('Odds Ratio (Values > 1 increase churn hazard)')
plt.legend()
plt.show()
```
