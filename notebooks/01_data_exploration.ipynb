# 01. Exploratory Data Analysis (EDA)
## E-Commerce Customer Retention & Churn Analytics
This notebook performs initial exploratory data analysis on the raw relational e-commerce tables: customers, orders, order items, products, payments, and marketing channels.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
RAW_DIR = Path('../data/raw')

customers = pd.read_csv(RAW_DIR / 'customers.csv')
orders = pd.read_csv(RAW_DIR / 'orders.csv')
products = pd.read_csv(RAW_DIR / 'products.csv')
print(f'Customers: {len(customers):,}')
print(f'Orders: {len(orders):,}')
print(f'Products: {len(products):,}')
```

### Customer Age & Demographic Distribution


```python
plt.figure(figsize=(10, 4))
sns.histplot(customers['age'], bins=25, kde=True, color='royalblue')
plt.title('Customer Age Distribution (Mean ~ 35 Years)')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

### Order Value Distributions and Outliers


```python
plt.figure(figsize=(10, 4))
sns.boxplot(x=orders['net_amount'], color='mediumaquamarine')
plt.title('Order Net Amount Boxplot (Distribution & Outliers)')
plt.xlabel('Net Amount ($)')
plt.show()
```
